##How to filter data



*   Using slicing method



```python
df[df["mark"]>90]
```

```python
df[df["class"]=="Six"]
```



*   Boolean filter



```python
df["mark"] >90
#return true false
```

```python
df[(df["class"]=="Six") & (df["gender"]=="male")]
```

```python
df[(df["gender"]=="female") & (df["mark"]>=85)]
```

```python
df[(df["gender"]=="female") | (df["mark"]>=90)]
# | this is called pipe
```

```python
df.loc[df["gender"]=="male", ["name","class"]]
```

```python
df.groupby("gender")["mark"].max()
# Max mark for male and max mark for female students
```

```python
df[df["class"] == "five"]["mark"].mean()
```

```python
6 - df[(df["class"] == "Six") & (df["gender"] == "male")]["mark"].mean()
#Average mark for class 'Six' and male students only
```

## How to sort values

```python
df.sort_values(by="mark", ascending=False)
#sort values by mark desc order
```

```python
df[["name", "class", "mark"]].sort_values(by="class")
```

```python
df.loc[df["mark"]>50, :].sort_values(by="mark")
```
